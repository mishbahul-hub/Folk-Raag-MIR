```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
PATH = "/kaggle/input/datasets/mishbhaul/wallpapers/black-hole-event-3840x2160-10324.jpg"
```

#### Applying a **Fourier Transform** to an image works the same core idea as with audio — just extended to 2D. Instead of decomposing a 1D signal (amplitude over time) into frequencies, you decompose a 2D signal (pixel intensity over x and y) into **spatial frequencies** — how rapidly brightness changes across the image, in both directions.

```python
img = cv2.imread(PATH, cv2.IMREAD_GRAYSCALE)
f = np.fft.fft2(img)
f_shifted = np.fft.fftshift(f)
magnitude_spectrum = 20 * np.log(np.abs(f) + 1) 
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(img, cmap='gray')
axes[0].set_title('Original Image')
axes[1].imshow(magnitude_spectrum, cmap='gray')
axes[1].set_title('Fourier Spectrum')
plt.show()
```
